````python
import os, re, unicodedata
import pandas as pd

def _normcols(s):
    s = str(s).strip()
    s = unicodedata.normalize("NFKD", s).encode("ascii","ignore").decode("ascii")
    return s.lower()

def _pick_ciiu_col(df):
    cols = df.columns.tolist()
    name_cands = [c for c in cols if re.search(r'(ciiu|rama4d|cod.*cii|clase.*eco|actividad)', c, flags=re.I)]
    cands = name_cands if name_cands else cols
    best, best_score = None, -1
    for c in cands:
        s = df[c].astype(str).str.extract(r'(\d{3,6})')[0]
        s = s.dropna().str.strip().apply(lambda x: x[:4] if len(x)>=4 else x.zfill(4))
        score = s.str.fullmatch(r'\d{4}').sum()
        if score > best_score:
            best, best_score = c, score
    return best, best_score

col_ciiu, score = _pick_ciiu_col(df_empleo)
if col_ciiu is None or score == 0:
    raise KeyError("No se encontró en df_empleo una columna fuente para construir 'cod_ciiu'.")

tmp = df_empleo[col_ciiu].astype(str).str.extract(r'(\d{3,6})')[0]
df_empleo['cod_ciiu'] = tmp.fillna('').str.strip().apply(lambda x: x[:4] if len(x)>=4 else x.zfill(4)).replace({'': pd.NA})

base_path = "/content/drive/MyDrive/datasets/geih_2025"
file_path = os.path.join(base_path, "2025-01-14 Correlativa CIIU Rev4.xlsx")

ciiu = pd.read_excel(file_path, sheet_name="CIIU 2022", dtype=str)
ciiu.columns = [_normcols(c) for c in ciiu.columns]

code_cands = [c for c in ciiu.columns if 'rama4d' in c or ('cod' in c and 'cii' in c)]
if not code_cands:
    for c in ciiu.columns:
        vals = ciiu[c].astype(str).str.strip()
        if len(vals.dropna()) and (vals.str.fullmatch(r'\d{4}').sum() > 0.3*len(vals.dropna())):
            code_cands.append(c); break
if not code_cands:
    raise KeyError("No se encontró columna de código CIIU (4 dígitos) en 'CIIU 2022'.")
code_col = code_cands[0]

desc_cands = [c for c in ciiu.columns if ('desc' in c or 'descrip' in c)]
if not desc_cands:
    lens = {c: ciiu[c].astype(str).str.len().mean() for c in ciiu.columns}
    desc_col = max(lens, key=lens.get)
else:
    desc_col = desc_cands[0]

ciiu = ciiu[[code_col, desc_col]].copy()
ciiu[code_col] = ciiu[code_col].astype(str).str.strip().str.extract(r'(\d{3,6})')[0].fillna('')
ciiu[code_col] = ciiu[code_col].apply(lambda x: x[:4] if len(x)>=4 else x.zfill(4))
ciiu[desc_col] = ciiu[desc_col].astype(str).str.strip()

merged = df_empleo.merge(ciiu, left_on='cod_ciiu', right_on=code_col, how='left')
merged = merged.rename(columns={desc_col: 'DESCRIPCION_CIIU'})
nuevas = set(merged.columns) - set(df_empleo.columns)
extra = [c for c in nuevas if c != 'DESCRIPCION_CIIU']
df_empleo = merged.drop(columns=extra, errors='ignore')
````

# Ejercicio 9: Groupby (20%)

Agrupar los datos por categorías, nos permite realizar análisis más profundos y obtener insights valiosos, que a simple vista no vemos en la base de datos.

1. Crea un nuevo dataframe llamado "df_sexo" donde se agrupe por 'sexo' y se calcule la suma de las columnas 'poblacion_total_ex' y 'ocupados_ex'. Crea una nueva columna que se llame 'porcentaje_ocupados' donde dividas las columna 'ocupados_ex' sobre 'poblacion_total_ex'. Al final usa reset_index() para convertir las categorías de sexo en una columna.

````python
### TU CÓDIGO AQUÍ ###
df_sexo = df_empleo.groupby('sexo')[['poblacion_total_ex', 'ocupados_ex']].sum()

df_sexo['porcentaje_ocupados'] = df_sexo['ocupados_ex'] / df_sexo['poblacion_total_ex']

df_sexo = df_sexo.reset_index()

display(df_sexo)
````
